### Snow data analysis
This notebook reproduces the analysis for the snow data, based on the results obtained by running `fit_conditional.py`. 

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import bootstrap
df10 = pd.read_json("results/2019-10-perturbed.json")
df20 = pd.read_json("results/2019-20-perturbed.json")
df30 = pd.read_json("results/2019-30-perturbed.json")
df50 = pd.read_json("results/2019-50-perturbed.json")
# uncomment to get the results for the unperturbed data
# df10 = pd.read_json("results/2019-10.json")
# df20 = pd.read_json("results/2019-20.json")
# df30 = pd.read_json("results/2019-30.json") 
# df50 = pd.read_json("results/2019-50.json")


```

```python
for d in [df10, df20, df30, df50]:
    res = bootstrap(
        (np.sqrt(d["GPFITC"][d["sCMOGP"] >0]) - np.sqrt(d["sCMOGP"][d["sCMOGP"] >0]),),
        lambda x: np.median(x),
        n_resamples=10000,
        random_state=42,
        method='percentile'
    )
    median_diff = np.median(np.sqrt(d["GPFITC"][d["sCMOGP"] >0]) - np.sqrt(d["sCMOGP"][d["sCMOGP"] >0]),)
    print(np.median(np.sqrt(d["sCMOGP"])), np.median(np.sqrt(d["GPFITC"])))
    print(f"median diff: {median_diff:.4f}")
    print(f"ci: [{res.confidence_interval.low:.4f}, {res.confidence_interval.high:.4f}]")
    print(np.median(np.sqrt(d[d["sCMOGP"] > 0]), axis=0))
    p_value = 2 * min(
    np.mean(res.bootstrap_distribution <= 0), 
    np.mean(res.bootstrap_distribution >= 0)
    )
    print(p_value)
```
