### Analysis of soil moisture results
This notebook reproduces (most of) Table 1 in the main paper.

```python
import os 
import numpy as np
import pandas as pd
for result in os.listdir("results"):
    print(result)
    df = pd.read_json("results/"+result)
    df.index = [5, 10, 50]
    df = np.sqrt(df).transpose()
    formatted_df = df.style.format( precision=3)
    print(formatted_df.to_latex())
```
